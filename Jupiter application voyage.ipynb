##  1. Présentation du Projet
Ce projet permet d'automatiser la création d'un carnet de voyage personnalisé. À partir d'une ville de départ et d'une destination, l'application s'occupe de tout : calcul du trajet, sélection d'un hôtel, choix de restaurants et suggestions d'activités culturelles.

L'objectif est d'offrir une **aide à la décision** grâce à :
* La centralisation de données provenant de sources multiples.
* Un calcul de budget prévisionnel précis.
* Une comparaison visuelle et tarifaire entre deux destinations possibles.

<a name="partie-4"></a>
## 4️⃣ Partie 4 : Activités & Programme quotidien

Cette section télécharge les données culturelles (API Ministère de la Culture) à proximité de **{destination}**.
Le programme est ensuite généré automatiquement en répartissant :
* Une activité le matin et une en soirée.
* Un restaurant le midi et un le soir.
* Le calcul de la météo pour chaque jour du séjour.

```python
# --- 1. RÉCUPÉRATION DES ACTIVITÉS ---
# Interroge l'API pour trouver des musées, monuments, parcs, etc.
activites_liste = mon_voyage.recuperer_activites()

# --- 2. GÉNÉRATION DU PLANNING ---
# Organise les journées et calcule le budget final (activités incluses)
mon_voyage.generer_programme(activites_liste)

# --- 3. EXPORT DES DONNÉES ---
# Sauvegarde vos listes en fichiers CSV pour une consultation externe
mon_voyage.exporter_hotels_csv()
mon_voyage.exporter_restos_csv()
mon_voyage.exporter_activites_csv()
```

<a name="part-5"></a>
## 5️⃣ Partie 5 : Budget Final & Cartographie

Dernière étape du projet :
* **Bilan Financier** : Récapitulatif du coût total par rapport au budget initial.
* **Carte Interactive** : Visualisation de la destination, des hôtels et des points d'intérêt culturels.

```python
# --- 1. RÉCUPÉRATION DES COORDONNÉES ---
lat, lon = mon_voyage.collecte.get_localisation(destination)

if lat and lon:
    # --- 2. CRÉATION DE LA CARTE ---
    ma_carte = CartesVoyage(destination, lat, lon)

    # --- 3. AJOUT DES ÉLÉMENTS ---
    # Ajoute les activités récupérées en partie 4
    ma_carte.ajouter_activites(activites_liste)
    
    # Ajoute les hôtels (récupération des données brutes pour les coordonnées)
    hotels_bruts = mon_voyage.collecte.collecter_hotels_carte(destination)
    ma_carte.ajouter_hotels(hotels_bruts)

    # --- 4. SAUVEGARDE ET AFFICHAGE ---
    ma_carte.sauvegarder("carte_mon_voyage.html")
    display(ma_carte.carte)
else:
    print("❌ Impossible d'afficher la carte : localisation introuvable.")
```

<a name="partie-6"></a>
## 6️⃣ Partie 6 : Optimisation & Comparaison

Cette section vous permet de personnaliser davantage votre séjour :
* **Changement d'Hôtel** : Si la sélection aléatoire ne vous convient pas, choisissez-en un autre parmi la liste.
* **Comparateur de Destination** : Comparez le budget de votre voyage actuel avec une autre ville (ex: Tours ou Blois) pour voir si vous pourriez économiser.

```python
# --- OPTIONNEL : CHANGER D'HÔTEL ---
# Cette méthode affiche la liste et demande une saisie utilisateur
# Si l'utilisateur appuie sur Entrée sans numéro, l'hôtel actuel est conservé.
mon_voyage.changer_hotel() 

# --- COMPARATIF ET CARTE VILLE ALTERNATIVE --- <a name="partie-6"></a>
print("\n" + "="*50)
print("ANALYSE COMPARATIVE")
print("="*50)

# 1. Exécution du comparatif budgétaire (affiche le tableau comparatif)
# La méthode demande le nom de la ville via un input()
mon_voyage.comparer_ville() 

# 2. Récupération du nom de la ville saisie pour la carte
# On extrait le nom de la ville pour laquelle on veut la carte
autre_ville = input("\nConfirmez le nom de la ville pour afficher sa carte : ")

# 3. Génération visuelle pour la ville alternative
print(f"\n📍 Visualisation géographique de {autre_ville.upper()}...")
lat_alt, lon_alt = mon_voyage.collecte.get_localisation(autre_ville)

if lat_alt and lon_alt:
    # Initialisation de la carte alternative
    carte_alt = CartesVoyage(autre_ville, lat_alt, lon_alt)
    
    # Collecte et ajout des activités
    act_alt = mon_voyage.collecte.collecter_activites(autre_ville)
    carte_alt.ajouter_activites(act_alt)
    
    # Affichage de la carte
    display(carte_alt.carte)
else:
    print("Impossible de générer la carte : ville non localisée.")
```
